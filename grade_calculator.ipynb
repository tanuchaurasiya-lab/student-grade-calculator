```python
def calculate_grade(marks):
    """Determines the grade and encouraging the message based on marks (0-100)"""
    if 90 <= marks <=100:
        return "A", "Outstanding! Excellent work! 🌟"
    elif 80 <= marks <=89:
        return "B", "Very Good! Keep it up! 👍"
    elif 70 <= marks <=79:
        return "C", "Good job! You're doing well, room to grow! 🌱"
    elif 60 <= marks <=69:
        return "D", "Needs improvements. Keep practicing."
    else:
        return "F", "Don't give up! let's review and try again."
    
def main():
    print("---------------------------------------")
    print("       STUDENT GRADE CALCULATOR        ")
    print("---------------------------------------")
    
    #Get the student name
    name = input("Enter student name:  ").strip()
    if not name:
        name = "Student"
        
    #Input validation loop for marks (0-100)
    while True:
        try:
            marks = float(input("Enter marks (0-100):  "))
            if 0 <= marks <= 100:
                break
            else: 
                print("❌ Invalid range! Marks must be between (0 and 100). Try again.\n")    
        except ValueError:
            print("❌ Invalid input! Please enter a valid number.\n")
            
    #Calculate grade and retrieve message 
    grade, message= calculate_grade(marks)
    
    #Display formatted result   
    print("\n" + "="*40)
    print(f"📊 RESULT FOR {name.upper()}:")
    print(f"Marks: {marks}/100")
    print(f"Grade: {grade}")
    print(f"Message: {message}")
    print("="*40) 
    
if __name__ == "__main__":
    main()
```
